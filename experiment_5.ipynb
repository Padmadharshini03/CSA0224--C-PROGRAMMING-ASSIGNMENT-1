```python
# Program 5: K-Nearest Neighbours (K-NN)
import numpy as np
from collections import Counter
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier


class KNN:
    def __init__(self, k=3):
        self.k = k

    def fit(self, X, y):
        self.X_train = np.array(X)
        self.y_train = np.array(y)

    def _predict_one(self, x):
        # Euclidean distance from x to every training sample
        distances = np.sqrt(np.sum((self.X_train - x) ** 2, axis=1))
        # indices of the k nearest neighbours
        nearest = np.argsort(distances)[:self.k]
        labels = self.y_train[nearest]
        # majority vote
        return Counter(labels).most_common(1)[0][0]

    def predict(self, X):
        return np.array([self._predict_one(x) for x in np.array(X)])


# Load data and split into training and test sets
iris = load_iris()
X, y = iris.data, iris.target
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y)

# Our own implementation
knn = KNN(k=3)
knn.fit(X_train, y_train)
y_pred = knn.predict(X_test)

print("Results of K-NN (k = 3) implemented from scratch")
print("-" * 55)
for i in range(len(y_test)):
    status = "Correct" if y_pred[i] == y_test[i] else "WRONG"
    print(f"Actual: {iris.target_names[y_test[i]]:10s} "
          f"Predicted: {iris.target_names[y_pred[i]]:10s} {status}")

accuracy = np.mean(y_pred == y_test) * 100
print(f"\nAccuracy (own K-NN): {accuracy:.2f}%")

# Compare with scikit-learn
sk_knn = KNeighborsClassifier(n_neighbors=3)
sk_knn.fit(X_train, y_train)
print(f"Accuracy (scikit-learn K-NN): {sk_knn.score(X_test, y_test) * 100:.2f}%")

# Effect of k
print("\nAccuracy for different values of k:")
for k in [1, 3, 5, 7, 9]:
    model = KNN(k=k)
    model.fit(X_train, y_train)
    acc = np.mean(model.predict(X_test) == y_test) * 100
    print(f"  k = {k}: {acc:.2f}%")

# Classify a new sample
new_sample = np.array([[5.1, 3.5, 1.4, 0.2]])
pred = knn.predict(new_sample)[0]
print("\nNew sample", new_sample[0], "-> predicted class:", iris.target_names[pred])
```
