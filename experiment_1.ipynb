```python
# Save the raw CSV data to a file
csv_data = """Sky,AirTemp,Humidity,Wind,Water,Forecast,EnjoySport
Sunny,Warm,Normal,Strong,Warm,Same,Yes
Sunny,Warm,High,Strong,Warm,Same,Yes
Rainy,Cold,High,Strong,Warm,Change,No
Sunny,Warm,High,Strong,Cool,Change,Yes"""

with open('data.csv', 'w') as f:
    f.write(csv_data)

print("Saved data to 'data.csv'")
```

```python
# Program 1: FIND-S algorithm
import pandas as pd

# Load the training data from the correct file name
data = pd.read_csv('data.csv')
print("Training data:")
print(data)
print()

attributes = data.iloc[:, :-1].values   # all columns except the last
target = data.iloc[:, -1].values        # last column = class label

# Step 1: initialise h to the most specific hypothesis
hypothesis = ['0'] * attributes.shape[1]
print("Initial hypothesis:", hypothesis)
print()

# Step 2: for each positive example, generalise h as needed
for i, row in enumerate(attributes):
    if str(target[i]).strip().lower() == 'yes':
        for j, value in enumerate(row):
            if hypothesis[j] == '0':
                hypothesis[j] = value
            elif hypothesis[j] != value:
                hypothesis[j] = '?'
        print(f"After example {i + 1} (positive): {hypothesis}")
    else:
        print(f"After example {i + 1} (negative): {hypothesis}   <- ignored")

print()
print("Most specific hypothesis found by FIND-S:")
print(hypothesis)
```
